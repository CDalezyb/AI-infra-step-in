```python
import torch
```

```python
def generate_tensor(shape):
    return  torch.arange(torch.prod(torch.tensor(shape))).reshape(shape)
    
```

1. transpose 是将两个维度进行交换

```python
x = generate_tensor((2, 3, 4))
print("original shape:", x.shape)
print("original x:", x)
x_transpose = x.transpose(0, 1)
print("transposed x.shape",x_transpose.shape)  # 输出: (3, 2, 4)
print("transposed x:", x_transpose)
```
